# DATASET APPROACH - DEFINICIÓN DE VARIABLES
Este script genera la estructura base
La siguiente tabla define el esquema de almacenamiento tabular. Cada fila representa una captura fotográfica única y sus variables fisiológicas/temporales asociadas.

| Variable | Tipo | Descripción y Origen | Ejemplo |
| :--- | :--- | :--- | :--- |
| `id_captura` | String | **(Key)** Clave primaria. Generada por la concatenación del ID del sujeto y el número de ingesta. | P001_T02 |
| `id_participante` | String | Identificador anónimo del sujeto. Permite agrupar métricas por individuo. | P001 |
| `tiempo_transcurrido_min` | Int | Minutos exactos desde el primer trago hasta la captura. Controla la latencia de absorción. | 45 |
| `peso_kg` | Float | Masa corporal del sujeto. Extraído de consulta directa. Afecta la distribución de alcohol en sangre. | 78.5 |
| `genero_biologico` | Int | `0`: Femenino, `1`: Masculino. Factor de corrección metabólica. | 1 |
| `gramos_alcohol_trago` | Float | Masa de alcohol de la ingesta inmediata. Cálculo: `(Volumen ml * ABV * 0.8) / 100`. | 14.2 |
| `gramos_alcohol_acumulados` | Float | **(Target)** Sumatoria temporal del alcohol ingerido. Variable a predecir por el modelo de regresión. | 42.6 |
| `[Nueva Variable]` | - | *Espacio reservado para futuras covariables (ej. horas de sueño previo, índice de fatiga).* | - |

```python
import os
import pandas as pd

# 1. Definir rutas absolutas/relativas respecto al script
DATA_DIR = os.path.dirname(os.path.abspath(__file__))
CSV_PATH = os.path.join(DATA_DIR, "dataset.csv")

# 2. Definir las columnas del esquema analítico
COLUMNS = [
    "id_captura",
    "id_participante",
    "tiempo_transcurrido_min",
    "peso_kg",
    "genero_biologico",
    "gramos_alcohol_trago",
    "gramos_alcohol_acumulados"
]

def inicializar_dataset(overwrite=True):
    """Crea o resetea el archivo dataset.csv con el esquema correcto."""
    if os.path.exists(CSV_PATH) and not overwrite:
        print(f"[-] El archivo ya existe en {CSV_PATH}. No se realizaron cambios.")
        return

    # Crear dataframe vacío solo con las columnas
    df = pd.DataFrame(columns=COLUMNS)
    df.to_csv(CSV_PATH, index=False, encoding="utf-8")
    print(f"[+] dataset.csv creado/sobrescrito exitosamente en: {CSV_PATH}")

if __name__ == "__main__":
    # Cambiar a True para resetear el CSV cuando cambies el esquema de columnas
    inicializar_dataset(overwrite=True)
```
